2. Find the lowest-priced product.
3. Calculate the total quantity of products sold.
4. Calculate the total sales value.
5. Calculate the average selling price of products sold.
6. Find the number of orders for each order status.

SQL concepts: SUM(), AVG(), MIN(), MAX(), GROUP BY, HAVING.

Important business calculation:
𝑆𝑎𝑙𝑒𝑠𝑉𝑎𝑙𝑢𝑒 = 𝑄𝑢𝑎𝑛𝑡𝑖𝑡𝑦 × 𝑆𝑒𝑙𝑙𝑖𝑛𝑔𝑃𝑟𝑖𝑐𝑒

Students should calculate sales from the Order_Items table, not simply add the current
product prices.


```python
#Find the highest-priced product.
query = """
SELECT * FROM Products WHERE unit_price =(SELECT MAX(unit_price)FROM Products);
"""
pd.read_sql_query(query, conn)
```

```python
#Find the lowest-priced product.
query = """
SELECT * FROM Products WHERE unit_price =(SELECT MIN(unit_price)FROM Products);
"""
pd.read_sql_query(query, conn)
```

```python
#Calculate the total quantity of products sold.
query = """
SELECT SUM(oi.quantity) AS Total_Quantity_Sold FROM Order_Items oi JOIN Orders o ON oi.order_id = o.order_id WHERE o.order_status = 'Delivered';
"""
pd.read_sql_query(query, conn)
```

```python
#Calculate the total sales value.
query = """
SELECT SUM(quantity * selling_price)AS Total_Sales_Value FROM Order_Items;
"""
pd.read_sql_query(query, conn)
```

```python
#Calculate the average selling price of products sold.
query = """
SELECT ROUND(AVG(selling_price),2)AS AVG_sellingprice FROM Order_Items;
"""
pd.read_sql_query(query, conn)
```

```python
#Find the number of orders for each order status.
query = """
SELECT order_status,COUNT(*)FROM Orders GROUP BY order_status"""
pd.read_sql_query(query, conn)
```

**Mission 3 – Customer behaviour analysis**

Management question: Who are QuickCart's most valuable customers?

1. Display each order along with the customer name.
2. Find the number of orders placed by each customer.
3. Calculate the total amount spent by each customer.
4. Identify the customer who has spent the most.
5. Find customers who have never placed an order.
6. Find customers who have placed more than two orders.
7. Find the average order value.
8. Identify the city with the highest number of customers.
9. Find customers who registered before a specified date.
10. Identify repeat customers.

SQL concepts: INNER JOIN, LEFT JOIN, GROUP BY, HAVING, subqueries.


```python
#Display each order along with the customer name.
query = """
SELECT o.order_id,c.customer_name FROM Orders o JOIN Customers c ON o.customer_id = c.customer_id"""
pd.read_sql_query(query, conn)
```

```python
#Find the number of orders placed by each customer.
query = """
SELECT c.customer_name,COUNT(o.order_id) AS Total_Orders FROM Customers c
LEFT JOIN Orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id;"""
pd.read_sql_query(query, conn)
```

```python
#Calculate the total amount spent by each customer.
query = """
SELECT c.customer_name, SUM(oi.quantity * oi.selling_price) AS Total_Amount_Spent
FROM Customers c
JOIN Orders o
ON c.customer_id = o.customer_id
JOIN Order_Items oi
ON o.order_id = oi.order_id
GROUP BY c.customer_id"""
pd.read_sql_query(query, conn)
```

```python
#Identify the customer who has spent the most.
query = """
SELECT c.customer_name,MAX(oi.selling_price)FROM Order_Items oi
JOIN Orders o ON oi.order_id= o.order_id
JOIN customers c ON o.customer_id = c.customer_id"""
pd.read_sql_query(query, conn)
```

```python
#Find customers who have never placed an order.
query = """
SELECT c.customer_name
FROM Customers c
LEFT JOIN Orders o
ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;"""
pd.read_sql_query(query, conn)
```

```python
#Find customers who have placed more than two orders.
query = """
SELECT c.customer_name, COUNT(o.order_id) AS Total_Orders FROM Customers c
INNER JOIN Orders o
ON c.customer_id = o.customer_id
GROUP BY c.customer_id
HAVING COUNT(o.order_id) > 2;"""
pd.read_sql_query(query, conn)
```

```python
#Find the average order value.
query = """
SELECT AVG(Order_Total) AS Average_Order_Value FROM
(SELECT order_id,SUM(quantity * selling_price) AS Order_Total FROM Order_Items GROUP BY order_id) AS Order_Values;"""
pd.read_sql_query(query, conn)
```

```python
#Identify the city with the highest number of customers.
query = """
SELECT city,COUNT(customer_id)AS Total_Customers FROM Customers ORDER BY Total_Customers """
pd.read_sql_query(query, conn)
```

```python
#Find customers who registered before a specified date.
query = """
SELECT * FROM Customers WHERE registration_date < '2025-03-20'"""
pd.read_sql_query(query, conn)
```

```python
#Identify repeat customers
query = """
SELECT c.customer_name,COUNT(o.order_id) AS Total_Orders FROM Customers c
LEFT JOIN Orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id HAVING Total_Orders > 1;"""
pd.read_sql_query(query, conn)
```

**Mission 4 – Payment and order investigation**

Management question: Are there any issues with order fulfilment and payments?

1. Display orders with their corresponding payment status.
2. Find all unpaid orders.
3. Find all cancelled orders.
4. Calculate the total amount received through each payment method.
5. Count successful and failed payments.
6. Find orders where the order status is Delivered but the payment status is not Paid.
7. Identify the most commonly used payment method.

SQL concepts: Multiple-table joins, filtering, conditional aggregation using CASE WHEN.


```python
#Display orders with their corresponding payment status.
query = """
SELECT o.order_id,o.customer_id, o.order_date,o.order_status,p.payment_status FROM Orders o
JOIN Payments p ON o.order_id = p.order_id;
"""
pd.read_sql_query(query, conn)
```

```python
#Find all unpaid orders.
query = """
SELECT o.order_id,o.customer_id, o.order_date,o.order_status,p.payment_status FROM Orders o
JOIN Payments p ON o.order_id = p.order_id
WHERE payment_status= 'Pending';
"""
pd.read_sql_query(query, conn)
```

```python
#Find all cancelled orders.
query = """
SELECT *FROM Orders WHERE order_status= 'Cancelled';
"""
pd.read_sql_query(query, conn)
```

```python
#Calculate the total amount received through each payment method.
query = """
SELECT payment_method,SUM(amount)FROM Payments GROUP BY payment_method;
"""
pd.read_sql_query(query, conn)
```

```python
#Count successful and failed payments.
query = """
SELECT payment_status,COUNT(payment_id)FROM payments GROUP BY payment_status;
"""
pd.read_sql_query(query, conn)
```

```python
#Find orders where the order status is Delivered but the payment status is not Paid.
query = """
SELECT o.order_id,o.customer_id, o.order_date,o.order_status,p.payment_status FROM Orders o
JOIN Payments p ON o.order_id = p.order_id
WHERE p.payment_status != 'Paid' AND o.order_status= 'Delivered';
"""
pd.read_sql_query(query, conn)
```

```python
#Identify the most commonly used payment method
query = """
 SELECT payment_method FROM Payments
 GROUP BY payment_method
 ORDER BY COUNT(payment_id) DESC LIMIT 1;
"""
pd.read_sql_query(query, conn)
```

**Mission 5 – Management challenge**

investigate:

1. Which product category should receive more attention from the marketing team?
2. Which products need restocking?
3. Which payment method is most popular?
4. Find the second-highest-priced product
5. Find products that have never been ordered
6. Find orders containing more than one product line

```python
#Which product category should receive more attention from the marketing team?
query = """
 SELECT p.category,
    SUM(oi.quantity) AS Total_Quantity_Sold,
    SUM(oi.quantity * oi.selling_price) AS Total_Sales
FROM Products p
JOIN Order_Items oi
ON p.product_id = oi.product_id
JOIN Orders o
ON oi.order_id = o.order_id
WHERE o.order_status = 'Delivered'
GROUP BY p.category
ORDER BY Total_Sales ASC LIMIT 1;
"""
pd.read_sql_query(query, conn)
```

```python
#Which products need restocking?
query = """
 SELECT * FROM Products WHERE stock_quantity =(SELECT MIN(stock_quantity)FROM Products);
"""
pd.read_sql_query(query, conn)
```

```python
#Which payment method is most popular?
query = """
 SELECT payment_method FROM Payments
 GROUP BY payment_method
 ORDER BY COUNT(payment_id) DESC LIMIT 1;
"""
pd.read_sql_query(query, conn)
```

```python
#Find the second-highest-priced product
query = """
 SELECT *FROM Products ORDER BY unit_price DESC LIMIT 1 OFFSET 1;"""
pd.read_sql_query(query, conn)
```

```python
#Find products that have never been ordered
query = """
SELECT p.product_name
FROM Products p
LEFT JOIN Order_Items oi
ON p.product_id = oi.product_id
WHERE oi.order_id IS NULL;
"""
pd.read_sql_query(query, conn)
```

```python
#Find orders containing more than one product line
query = """
 SELECT order_id,COUNT(order_id)AS product_lines FROM Order_Items
 GROUP BY Order_Id HAVING product_lines >1;
"""
pd.read_sql_query(query, conn)
```
